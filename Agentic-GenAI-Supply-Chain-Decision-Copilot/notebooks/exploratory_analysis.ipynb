# Exploratory Data Analysis — Supply Chain Demand Patterns

This notebook performs exploratory analysis on historical sales data to identify demand distributions, category trends, day-of-week patterns, and seasonality.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load raw sales dataset
data_path = os.path.join("..", "data", "raw", "sales.csv")
sales_df = pd.read_csv(data_path)
sales_df['sale_date'] = pd.to_datetime(sales_df['sale_date'])
print(f"Total Sales Records: {len(sales_df)}")
sales_df.head()
```

## 1. Overall Sales Demand Analysis

```python
daily_total = sales_df.groupby('sale_date')['quantity'].sum().reset_index()
plt.figure(figsize=(12, 4))
plt.plot(daily_total['sale_date'], daily_total['quantity'], color='teal', linewidth=1)
plt.title('Overall Daily Demand Across All Warehouses')
plt.xlabel('Date')
plt.ylabel('Units Sold')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## 2. Weekly & Monthly Seasonality

```python
sales_df['day_name'] = sales_df['sale_date'].dt.day_name()
sales_df['month_name'] = sales_df['sale_date'].dt.month_name()

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
order_days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
sns.barplot(data=sales_df, x='day_name', y='quantity', order=order_days, ax=axes[0], palette='Blues_d')
axes[0].set_title('Average Demand by Day of Week')

sns.barplot(data=sales_df, x='month_name', y='quantity', ax=axes[1], palette='Viridis')
axes[1].set_title('Average Demand by Month')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 2. Generate and Plot 14-Day Demand Forecast for Sample Product

```python
forecast_res = forecast_product(product_id=1, horizon_days=14)
f_df = pd.DataFrame(forecast_res['forecast'])
f_df['date'] = pd.to_datetime(f_df['date'])

plt.figure(figsize=(10, 4))
plt.plot(f_df['date'], f_df['predicted_demand'], marker='o', color='indigo', label='XGBoost Predicted Demand')
plt.fill_between(f_df['date'], f_df['lower_bound'], f_df['upper_bound'], color='indigo', alpha=0.15, label='Uncertainty Bound (MAE)')
plt.title(f"14-Day Demand Forecast for Product ID 1 ({forecast_res['sku']})")
plt.xlabel('Forecast Date')
plt.ylabel('Predicted Units')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
